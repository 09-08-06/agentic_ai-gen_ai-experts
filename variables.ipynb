# variables in python 

```python
first_name = 'tanvi'
last_name = 'vijay'
country = 'india'
city = 'indore'
age = 19
is_married = False
skills = ['html','css','js']
```

```python
# printing the values stored in variables
print('First Name : ',first_name)
print('Last Name : ',last_name)
print('Country : ',country)
print('City : ',city)
print('Age : ',age)
print('married : ',is_married)
print('skills : ',skills)
```

```python
# declaring multiple variables in one line
first_name,last_name,country,city,age,is_married,skills ='tanvi','vijay','india','indore',19,False,['html','css','js']
print(first_name , last_name , country , city , age , is_married , skills)
print('First Name : ',first_name)
print('Last Name : ',last_name)
print('Country : ',country)
print('City : ',city)
print('Age : ',age)
print('married : ',is_married)
print('skills : ',skills)
```
